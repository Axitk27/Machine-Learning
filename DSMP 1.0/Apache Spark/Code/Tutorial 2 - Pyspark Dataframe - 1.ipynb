```python
#drop columns 
df.drop('Name').show()
```

```python
#null value is droped - rows dropped where none value is present
df.na.drop().show()
```

```python
#how to replace
# how = any if any of the columns are having the none value then it will be dropped
df.na.drop(how='any').show()
```

```python
# how = any. if all of the columns are having the none value then it will be dropped
df.na.drop(how='all').show()
```

```python
#threshold 
# threshold = 2 means atlease 2 non value should be present
df.na.drop(how='any',thresh=2).show()
```

```python
# threshold = 1 means atlease 1 non value should be present
df.na.drop(how='any',thresh=1).show()
```

```python
# threshold = 3 means atlease 1 non value should be present
df.na.drop(how='any',thresh=3).show()
```

```python
#subset
# None value for specific column has been deleted
df.na.drop(how='any',subset=['Name']).show()
```

```python
#fill the missing value
#it is littel bit smart as it is also checking the type as well
df.na.fill('Missing Value',['Name','packages']).show()
```

```python
#fill with the mean value
df.na.fill(12,'cgpa').show()
```

```python
#Filter operation

df.filter("packages>=4").show()
```

```python
df.filter("placement==1").select(['Name','cgpa']).show()
```

```python
df.filter(~(df['packages']>=4) & (df['placement'] == 1)).show()
```

```python
#groupby and aggregate function
df = df.na.drop(subset=['Name'])
```

```python
df.groupby(['department']).sum('packages').show()
```

```python
df.groupby(['department']).count().show() 
```

```python
#this is the direct aggrigate function
df.agg({'packages':'max'}).show()
```
