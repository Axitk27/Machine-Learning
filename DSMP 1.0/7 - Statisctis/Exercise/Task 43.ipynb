```python
#(1) A company claims that their email marketing campaign has a 15% click-through rate. If you randomly select 100 people to receive the email, 
# what is the probability that exactly 20 will click through to the website?
p = 0.15
n = 100
c = 20

# It is a binomial tranformation

P = math.comb(n,c)*np.power(p,c)*np.power(1-p,n-c)
P
```
