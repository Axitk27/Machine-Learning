```python
import plotly.graph_objects as go
import numpy as np
import pandas as pd
import plotly.offline as pyo
import plotly.express as px
```

```python
match = pd.read_csv('/Dataset/matches.csv')
delivery = pd.read_csv('/Dataset/deliveries.csv')

delivery = delivery.merge(match,left_on='match_id',right_on='id')
match.head()
```

```python
top_50 = list(delivery['batsman'].value_counts().iloc[:50].index)
temp_df = delivery[delivery['batsman'].isin(top_50)]
batsman_runs = temp_df.groupby('batsman')['batsman_runs'].sum()
batsman_balls = temp_df.groupby('batsman')['batsman_runs'].count()



sr = 100*batsman_runs/batsman_balls
sr = sr.reset_index()
sr.rename(columns={'batsman_runs':'Strikerate'},inplace=True)
```

```python
temp_df = delivery[(delivery['batsman'].isin(top_50)) & ((delivery['player_dismissed'] == delivery['batsman']) | (delivery['player_dismissed'] == delivery['non_striker']))]
batsman_out = temp_df.groupby('batsman')['player_dismissed'].count().sort_values(ascending=False)
avg = (batsman_runs/batsman_out).reset_index()

avg = avg.merge(sr,on='batsman')
avg.rename(columns={0:'Average','batsman_runs':'Strikerate'},inplace=True)
avg.head()
```

```python
#plot a scatter plot

trace1 = go.Scatter(x=avg['Average'],y=avg['Strikerate'],mode='markers',text=avg['batsman'],marker={'color':'black','size':15})
data = [trace1]
layout = go.Layout(title='Average Vs Strikerate',xaxis={'title':'Batsman Average'},yaxis={'title':'Batsman Strikerate'})
fig = go.Figure(data=data,layout=layout)
# pyo.plot(fig)
fig.show()
```

```python
#line chart
#year by year virat's performance

temp_df = delivery[delivery['batsman'] == 'V Kohli']
teams_runs = temp_df.groupby('season')['batsman_runs'].sum().reset_index()

temp_df1 = delivery[delivery['batsman'] == 'MS Dhoni']
teams_runs1 = temp_df1.groupby('season')['batsman_runs'].sum().reset_index()

trace1 = go.Scatter(x=teams_runs['season'],y=teams_runs['batsman_runs'],mode='lines',text=avg['batsman'],marker={'color':'black','size':15},name='Virat Kohli')
trace2 = go.Scatter(x=teams_runs1['season'],y=teams_runs1['batsman_runs'],mode='lines',text=avg['batsman'],marker={'color':'black','size':15},name='MS Dhoni')

data = [trace1,trace2]
layout = go.Layout(title='Performance against all Teams',xaxis={'title':'Teams'},yaxis={'title':'Runs'})
fig = go.Figure(data=data,layout=layout)

fig.show()

```

```python
#bar chart

top_10 = list(delivery['batsman'].value_counts().iloc[:10].index)
temp_df = delivery[delivery['batsman'].isin(top_10)]
batsman_runs = temp_df.groupby('batsman')['batsman_runs'].sum().sort_values(ascending=False).reset_index()
batsman_runs
```

```python

trace1 = go.Bar(x=batsman_runs['batsman'], y=batsman_runs['batsman_runs'])
data = [trace1]
layout = go.Layout(title='Top 10 Batsman Runs',xaxis={'title':'Batsman'},yaxis={'title':'Batsman Runs'})
fig = go.Figure(data=data,layout=layout)
fig.show()
```

```python
#bubble chart
top_50 = list(delivery['batsman'].value_counts().iloc[:50].index)

temp_df = delivery[(delivery['batsman'].isin(top_50)) & ((delivery['player_dismissed'] == delivery['batsman']) | (delivery['player_dismissed'] == delivery['non_striker']))]
batsman_out = temp_df.groupby('batsman')['player_dismissed'].count().sort_values(ascending=False)

temp_df0 = delivery[(delivery['batsman'].isin(top_50))]
batsman_runs = temp_df0.groupby('batsman')['batsman_runs'].sum()

temp_df1 = delivery[(delivery['batsman'].isin(top_50)) & (delivery['batsman_runs'] == 6)]
six_number = temp_df1.groupby('batsman')['batsman_runs'].count().reset_index()
six_number.rename(columns = {'batsman_runs':'six'},inplace = True)


avg = (batsman_runs/batsman_out).reset_index()
final_frame = avg.merge(batsman_runs,on='batsman').merge(six_number,on='batsman')


final_frame.rename(columns={0:'Average','batsman_runs':'Runs'},inplace=True)

# avg.head()

```

```python
trace1 = go.Scatter(x=final_frame['Average'], y=final_frame['Runs'],mode='markers',marker={'size':final_frame['six']},text = final_frame['batsman'])
data = [trace1]
layout = go.Layout(title='Batsman Runs',
                   xaxis={'title':'Batsman Average'},
                   yaxis={'title':'Batsman Runs'})
fig = go.Figure(data=data,layout=layout)
fig.show()
```

```python
#box plot

# Column shsould be numerical and you will find a distribution of the plot
```

```python
match_agg = delivery.groupby('match_id')['total_runs'].sum().reset_index()
season_wise = match_agg.merge(match,left_on='match_id',right_on='id')[['match_id','total_runs','season']]
season_wise
```

```python
trace1 = go.Box(x=season_wise['total_runs'], name='All Season',marker={'color':'#00a65a'})
layout = go.Layout(title='Total Runs Box Plot',
                   xaxis={'title':'Total Score'},
                   )
data = [trace1]
fig = go.Figure(data=data,layout=layout)
fig.show()
```

```python
trace1 = go.Box(x=season_wise[season_wise['season']==2008], name='2008',marker={'color':'green'})
trace2 = go.Box(x=season_wise[season_wise['season']==2017], name='2017',marker={'color':'blue'})

data = [trace1,trace2]
layout = go.Layout(title='Total Runs Box Plot',
                   xaxis={'title':'Total Score'},
                   )
fig = go.Figure(data=data,layout=layout)
fig.show()
```

```python
#heatmaps

# both categorical or categorical and Numerical but third column should be numerical
```

```python
six = delivery[delivery['batsman_runs']==6]
six = six.groupby(['batting_team','over'])['batsman_runs'].count().reset_index()
six.rename(columns={'batsman_runs':'Six'},inplace=True)
```

```python
dots = delivery[delivery['batsman_runs']==0]
dots = dots.groupby(['batting_team','over'])['batsman_runs'].count().reset_index()
dots.rename(columns={'batsman_runs':'Dots'},inplace=True)
dots
```

```python
trace= go.Heatmap(x=six['batting_team'],y=six['over'],z=six['Six'])
data=[trace]

layout=go.Layout(title='Heat Map',xaxis={'title':'Batting Team'},yaxis={'title':'Six'})

fig=go.Figure(data=data,layout=layout)

fig.show()
```

```python
trace1= go.Heatmap(x=six['batting_team'],y=six['over'],z=six['Six'])
trace2= go.Heatmap(x=dots['batting_team'],y=dots['over'],z=dots['Dots'])

from plotly import tools
data=[trace1,trace2]

layout=go.Layout(title='Heat Map',xaxis={'title':'Batting Team'})

fig=tools.make_subplots(rows=1,cols=2,subplot_titles=["6's","0's"],shared_yaxes=True)
fig.append_trace(trace1,1,1)
fig.append_trace(trace2,1,2)

fig.show()
```
