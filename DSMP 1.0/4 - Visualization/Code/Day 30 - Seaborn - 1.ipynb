### 1. Relational Plot

- to see the statistical relation between 2 or more variables.
- Bivariate Analysis

Plots under this section

- scatterplot
- lineplot

```python
import seaborn as sns
import matplotlib.pyplot as plt
import plotly.express as px
```

```python
tips = sns.load_dataset('tips')
tips.head()
```

```python
# scatter plot -> axes level function
#this is a axis level function
#axis level function rectangle shape will be given
sns.scatterplot(data = tips,x='total_bill',y='tip')
```

```python
# relplot -> figure level -> square shape
#figure level function square shape will be given
sns.relplot(data=tips,x='total_bill',y='tip',kind='scatter',hue='sex',style='timesize')
```

```python
# scatter using relplot -> size and hue

sns.relplot(data=tips,x='total_bill',y='tip',kind='scatter',hue='sex')
```

```python
# style semantics

sns.relplot(data=tips,x='total_bill',y='tip',kind='scatter',hue='sex',style='time',size='size')
```

```python
# line plot
gap = px.data.gapminder()
gap.head()
```

```python
# axes level function
temp_df = gap[gap['country'] == 'India']
sns.lineplot(temp_df,x='year',y='lifeExp')
```

```python
# using relpplot
sns.relplot(data=temp_df,x='year',y='lifeExp',kind='lines')
```

```python
# hue -> style -> figure level has a more control compare to the axis level function
temp_df = gap[gap['country'].isin(['India','Pakistan','China'])]
sns.relplot(temp_df,x='year',y='lifeExp',kind='line',hue='country')
```

```python
temp_df = gap[gap['country'].isin(['India','Brazil','Germany','Poland','Sweden'])]
sns.relplot(temp_df,x='year',y='lifeExp',kind='line',hue='country',style='continent')
```

```python
# facet plot -> figure level function -> work with relplot
# only with figure level plot
# it will not work with scatterplot and lineplot function

sns.relplot(data=tips,x='total_bill',y='tip',col='sex',kind='scatter')
```

```python
# row wrap
sns.relplot(data=tips,x='total_bill',y='tip',row='sex',kind='scatter')
```

```python
sns.relplot(data=tips,x='total_bill',y='tip',row='sex',kind='scatter',col='smoker')
```

```python
sns.relplot(data=gap,x='lifeExp',y='gdpPercap',kind='scatter',col='year',col_wrap=3)
```

### 2. Distribution Plots

- used for univariate analysis
- used to find out the distribution
- Range of the observation
- Central Tendency
- is the data bimodal?
- Are there outliers?

Plots under distribution plot

- histplot
- kdeplot
- rugplot

```python
# figure level -> displot
# axes level -> histplot -> kdeplot -> rugplot
```

```python
# plotting univariate histogram

sns.histplot(data=tips,x='tip')
```

```python
sns.displot(tips,x='tip',kind='hist')
```

```python
# bins parameter
sns.displot(tips,x='tip',kind='hist',bins=5)
```

```python
# It’s also possible to visualize the distribution of a categorical variable using the logic of a histogram.
# Discrete bins are automatically set for categorical variables
sns.displot(data=tips,x='day',kind='hist')
# countplot - > both are same
sns.countplot(data=tips,x='day')
```

```python
# hue parameter
sns.displot(tips,x='tip',kind='hist',hue='sex')
```

```python
# element -> step
sns.displot(tips,x='tip',kind='hist',hue='sex',element='step')
```

```python
titanic = sns.load_dataset('titanic')
titanic
```

```python
sns.displot(data=titanic,x='age',element='step',hue='sex')
```

```python
# faceting using col and row -> not work on histplot function
sns.displot(data=titanic,x='age',element='step',col='sex')
sns.displot(data=titanic,x='age',element='step',row='sex')
```

```python
# kdeplot
# Rather than using discrete bins, a KDE plot smooths the observations with a Gaussian kernel, producing a continuous density estimate
sns.displot(data=titanic,x='age',kind='kde')
```

```python
sns.kdeplot(data=titanic,x='age')
```

```python
# hue -> fill
sns.kdeplot(data=titanic,x='age',hue='sex',fill=True)
```

```python
# Rugplot
# sns.kdeplot(data=titanic,x='age',fill=True)
sns.rugplot(data=titanic,x='age') #this will giv eyou the probablity of the data
# Plot marginal distributions by drawing ticks along the x and y axes.

# This function is intended to complement other plots by showing the location of individual observations in an unobtrusive way.
```

```python
# Bivariate histogram
# A bivariate histogram bins the data within rectangles that tile the plot
# and then shows the count of observations within each rectangle with the fill color

# sns.histplot(data=tips, x='total_bill', y='tip') #this looks like a box plot
sns.displot(data=tips, x='total_bill', y='tip',kind='hist') #this looks like a box plot
```

```python
# Bivariate Kdeplot
# a bivariate KDE plot smoothes the (x, y) observations with a 2D Gaussian
sns.displot(data=tips, x='total_bill', y='tip',kind='kde') #this looks like a box plot
```

### 3. Matrix Plot

- Heatmap
- Clustermap

```python
# Heatmap


# Plot rectangular data as a color-encoded matrix
data = gap.pivot(index='country',columns='year',values='lifeExp')
# axes level function
plt.figure(figsize=(15,15))
sns.heatmap(data=data)
```

```python
# annot
data = gap[gap['continent']=='Europe']
data = data.pivot(index='country',columns='year',values='lifeExp')
plt.figure(figsize=(10,10))
sns.heatmap(data=data,annot=True,cmap='viridis')
```

```python
# Clustermap

# Plot a matrix dataset as a hierarchically-clustered heatmap.

# This function requires scipy to be available.
iris = px.data.iris()
iris.head()
```

```python
sns.clustermap(iris.iloc[:,[0,1,2,3]]) #dendrogram
```
