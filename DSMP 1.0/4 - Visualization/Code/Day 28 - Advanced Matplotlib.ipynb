### 3D Surface Plots

```python
x = np.linspace(-10,10,100)
y = np.linspace(-10,10,100)
```

```python
xx,yy = np.meshgrid(x,y)
```

```python
z = xx**2 + yy**2
```

```python
fig = plt.figure(figsize = (15,10))

ax = plt.subplot(projection='3d')
graph = ax.plot_surface(xx,yy,z,cmap='viridis')
fig.colorbar(graph)
```

```python
# function 

z = np.sin(xx) + np.cos(yy)
fig = plt.figure(figsize = (15,10))

ax = plt.subplot(projection='3d')
graph = ax.plot_surface(xx,yy,z,cmap='viridis')
fig.colorbar(graph)
```

```python
#function 
z = np.tan(xx) + np.cos(yy)
fig = plt.figure(figsize = (15,10))

ax = plt.subplot(projection='3d')
graph = ax.plot_surface(xx,yy,z,cmap='viridis')
fig.colorbar(graph)
```

### Contour Plots

```python
#representing 3d graph into 2d plot

fig = plt.figure(figsize = (15,10))
z = xx**2 + yy**2
ax = plt.subplot()
graph = ax.contour(xx,yy,z,cmap='viridis')
fig.colorbar(graph)
```

```python
fig = plt.figure(figsize = (15,10))
z = xx**2 + yy**2
ax = plt.subplot()
graph = ax.contourf(xx,yy,z,cmap='viridis') #this is more easier comapre to the first one
fig.colorbar(graph)
```

### Heatmap

```python
delivery = pd.read_csv('Dataset\IPL_Ball_by_Ball_2008_2022.csv')
delivery.head()
```

```python
temp_df = delivery[(delivery['ballnumber'].isin([1,2,3,4,5,6])) & (delivery['batsman_run'] == 6)]

temp_df.head()
```

```python
grid = temp_df.pivot_table(index = 'overs',columns = 'ballnumber',values='batsman_run',aggfunc='count')
```

```python
plt.figure(figsize=(20,10))
plt.imshow(grid)
plt.yticks(temp_df['overs'].unique(),list(range(1,21)))
plt.xticks(list(np.arange(0,6)),list(range(1,7)))
plt.colorbar()
```
