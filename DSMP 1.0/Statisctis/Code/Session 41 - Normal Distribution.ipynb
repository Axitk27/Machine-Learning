```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('Data\iris.csv')
df1 = pd.read_csv('Data/titanic.csv')
```

```python
#PDF Plot

sns.displot(data = df,x = 'PetalLengthCm',hue = 'Species',kind = 'kde')
```

```python
#CDF Plot
sns.displot(data = df,x = 'PetalLengthCm',hue = 'Species',kind = 'kde')
sns.ecdfplot(data = df,x = 'PetalLengthCm',hue = 'Species')
```

```python
#2d Plot

sns.jointplot(data = df , x = 'PetalLengthCm', y = 'SepalLengthCm', kind = 'kde',fill = True,cbar= True)
```

```python
#3d Plot

sns.conturplot(data = df , x = 'PetalLengthCm', y = 'SepalLengthCm',z = 'PetalWidthCm', kind = 'kde',fill = True,cbar= True)
```
