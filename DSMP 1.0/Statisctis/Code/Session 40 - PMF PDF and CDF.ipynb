```python
import pandas as pd
import numpy as np
import random
```

```python

L = []

for i in range(1000):
    a = random.randint(1,6)
    b = random.randint(1,6)
    
    L.append(a+b)
```

```python
# Parametric Density Estimation

import matplotlib.pyplot as plt
from numpy.random import normal

sample = normal(loc=50,scale=5,size=1000)
```

```python
sample
```

```python
# Plot the histogram
sample.mean()
```

```python
#find the distribution of the plot

plt.hist(sample,bins=10)
plt.show()
```

```python
sample_mean = sample.mean()
sample_std = sample.std()
```

```python
# Fitting the distribution 

from scipy.stats import norm
dist = norm(sample_mean,sample_std)
```

```python
values = np.linspace(sample.min(),sample.max(),100)
```

```python
probabilities = [dist.pdf(value) for value in values]
```

```python
# Plot the histogram

plt.hist(sample,bins=10,density = True)
plt.plot(values,probabilities)
```

```python
import seaborn as sns
```

```python
sns.displot(sample,kind='kde')
```

```python
# Non Parametric Density Estimation 
```

```python
sample1 = normal(loc = 20,scale = 5,size = 300)
sample2 = normal(loc = 40,scale = 5,size = 700)
sample = np.hstack((sample1,sample2))
```

```python
plt.hist(sample,bins=50)
plt.show()
```

```python
from sklearn.neighbors import KernelDensity
model = KernelDensity(bandwidth = 3,kernel = 'gaussian')

#shape change to (1000,1)
sample = sample.reshape((len(sample),1))
model.fit(sample)
```

```python
values = np.linspace(sample.min(),sample.max(),100)
values = values.reshape((len(values),1))
```

```python
probablities = model.score_samples(values)
#First will give me the log of probabilities value
#Thats why I am doing here exp
probablities = np.exp(probablities)
```

```python
plt.hist(sample,bins=50,density=True)
plt.plot(values[:],probablities)
plt.show()
```

```python
sns.kdeplot(sample,bw_adjust=1)
```
