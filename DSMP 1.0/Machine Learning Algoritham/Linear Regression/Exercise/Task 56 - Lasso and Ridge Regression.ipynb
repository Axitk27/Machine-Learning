### `Task` Train a Lasso And Ridge Regression model on Boston House Price Datset. And also show how it reduces overfitting in compare to normal Linear regression model.

Data set Link - https://www.kaggle.com/datasets/altavish/boston-housing-dataset

DataSet description:
```
ZN: proportion of residential land zoned for lots over 25,000 sq.ft.
INDUS: proportion of non-retail business acres per town
CHAS: Charles River dummy variable (= 1 if tract bounds river; 0 otherwise)
NOX: nitric oxides concentration (parts per 10 million)
RM: average number of rooms per dwelling
AGE: proportion of owner-occupied units built prior to 1940
DIS: weighted distances to ﬁve Boston employment centers
RAD: index of accessibility to radial highways
TAX: full-value property-tax rate per $10,000
PTRATIO: pupil-teacher ratio by town 12. B: 1000(Bk−0.63)2 where Bk is the proportion of blacks by town 13. LSTAT: % lower status of the population
MEDV: Median value of owner-occupied homes in $1000s (Target)
```

````python
# Your Code goes here
import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression,Ridge,Lasso,ElasticNet
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score
from sklearn.preprocessing import OneHotEncoder
````

````python
df = pd.read_csv("D:\Education\Extra\Machine Learning\DSMP 1.0\Machine Learning Algoritham\Linear Regression\Exercise\Data\HousingData.csv")
df.head()
````

````python
df.isna().sum()
````

````python
df.dropna(inplace = True)
````

````python
df.isna().sum()
````

````python
x = df.iloc[:,:12]
y = df.iloc[:,12]
````

````python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=4,shuffle =True)
````

````python
#training a linear regression model

lr = LinearRegression()
ri = Ridge(alpha = 10)
la = Lasso(alpha = 0.1)
el = ElasticNet(alpha = 5,l1_ratio=0.01)
````

````python
lr.fit(x_train,y_train)
ri.fit(x_train,y_train)
la.fit(x_train,y_train)
el.fit(x_train,y_train)
````

````python
y_pred = lr.predict(x_train)
r2_score(y_train,y_pred)
````

````python
y_pred = ri.predict(x_train)
r2_score(y_train,y_pred)
````

````python
y_pred = la.predict(x_train)
r2_score(y_train,y_pred)
````

````python
y_pred_lr = lr.predict(x_test)
y_pred_ri = ri.predict(x_test)
y_pred_la = la.predict(x_test)
y_pred_el = el.predict(x_test)
````

````python
#score the r2 score - Linear regression

r2_score(y_test,y_pred_lr)
````

````python
#score the r2 score - Ridge regression
r2_score(y_test,y_pred_ri)
````

````python
#score the r2 score - Lasso regression

r2_score(y_test,y_pred_la)
````

````python
#score the r2 score - Elasticnet regression
r2_score(y_test,y_pred_el)
````

### `Task` perform Lasso and Ridge in Wine Quality dataset and show how it's different from Linear Regression.

Data link : https://docs.google.com/spreadsheets/d/e/2PACX-1vQDVwxneOKOaJL13QMhkAhYrgWlH1tICY7RacUnj_lL8m9uUWaaUf3p7bScNyh_D2Rvt7nc1q11adSy/pub?gid=647503637&single=true&output=csv



````python
# Loading Wine dataset
wine_data_path = "https://docs.google.com/spreadsheets/d/e/2PACX-1vQDVwxneOKOaJL13QMhkAhYrgWlH1tICY7RacUnj_lL8m9uUWaaUf3p7bScNyh_D2Rvt7nc1q11adSy/pub?gid=647503637&single=true&output=csv"
data = pd.read_csv(wine_data_path)
````

````python
data.shape
````

````python
data.isna().sum()
````

````python
data.dropna(axis=0,inplace = True)
````

````python
# Your Code goes here
data.head()
````

````python
encoder = OneHotEncoder(sparse_output=False)
one_hot_encoded = encoder.fit_transform(pd.DataFrame(data[["type"]]))
````

````python
x = data.iloc[:,1:]
y = one_hot_encoded
````

````python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=2)
````

````python
#train a Linear Regression model

lr = LinearRegression()
lr.fit(x_train,y_train)
y_pred = lr.predict(x_test)
print(r2_score(y_true=y_test,y_pred=y_pred))
````

````python
y_pred
````

````python
lr.coef_
````

````python
#train a Ridge Regression model

ri = Ridge(alpha = 0.01)
ri.fit(x_train,y_train)
y_pred = ri.predict(x_test)
print(r2_score(y_true=y_test,y_pred=y_pred))
````

````python
ri.coef_
````

````python
#train a Lasso Regression model

la = Lasso(alpha = 2)
la.fit(x_train,y_train)
y_pred = la.predict(x_test)
print(r2_score(y_true=y_test,y_pred=y_pred))
````
