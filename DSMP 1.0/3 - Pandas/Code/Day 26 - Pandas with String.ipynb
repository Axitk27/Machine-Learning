```python
# problem in vectorized opertions in vanilla python
s = ['cat','mat',None,'Rat']

[i.startswith('c') for i in s] #if you have a none data then python would not function on lthis data 
```

```python
# How pandas solves this issue?

s = pd.Series(['cat','mat',None,'Rat'])

# string accessor - str is called accessor
s.str.startswith('c')
# fast and optimized
```
