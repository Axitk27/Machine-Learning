```python
import numpy as np
import pandas as pd
```

```python
import requests
from bs4 import BeautifulSoup
```

```python
headers = {"User-Agent": "Mozilla/5.0 (Macintosh; Intel Mac OS X 10.14; rv:66.0) Gecko/20100101 Firefox/66.0"}
webpage = requests.get('https://www.scrapethissite.com/pages/simple/',headers=headers).text
```

```python
soup = BeautifulSoup(webpage,'lxml')
```

```python
soup = soup.prettify()
```

```python
soup.find('div',{'class':'container'}).find('div',{'class':'row'})
```

```python
country = soup.find('div',{'id':'page'}).find('section',{'id':'countries'}).find('div',{'class':'container'}).find_all('div',{'class':'col-md-4 country'})
```

```python
name = []
capital = []
population = []
area = []

for item in country:
    name.append(item.find('h3').text.strip())
    capital.append(item.find('div',{'class':'country-info'}).find('span',{'class':'country-capital'}).text.strip())
    population.append(item.find('div',{'class':'country-info'}).find('span',{'class':'country-population'}).text.strip())
    area.append(item.find('div',{'class':'country-info'}).find('span',{'class':'country-area'}).text.strip())
```

```python
df = pd.DataFrame({'Country':name,'Capital':capital,'Population':population,'Area':area})
```

```python
df.to_csv('Country')
```

```python
df['Area'] = df['Area'].astype('float')
```

```python
df['Area'].mean()
```

```python
df[df['Area'] > df['Area'].mean()].sort_values(by='Area',ascending = False)
```
